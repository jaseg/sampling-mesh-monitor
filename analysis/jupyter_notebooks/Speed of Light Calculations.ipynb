```python
import numpy as np
import math
import scipy.optimize
import scipy.constants
from matplotlib import pyplot as plt
#%matplotlib widget
```

```python
labels = ['PI3HDX12211', '74LVC1G157', 'MAX3748', 'TDP0604']
lengths = np.array([1.07, 1.93, 2.86, 3.86])
data = np.array([
    [16.9, 26.0, 36.4, 46.1],
    [17.1, 26.4, 36.6, 48.2],
    [17.2, 26.4, 36.6, 45.6],
    [17.0, 26.2, 36.5, 45.8]
])
```

```python
fig, ax = plt.subplots(figsize=(4, 2))

for series, label in zip(data, labels):
    ax.plot(lengths, series, label=label)
    f = lambda t, c, t_off: (t - t_off) * c
    (c, t_off), p_cov = scipy.optimize.curve_fit(f, series*1e-9, 2*lengths, (3e8/2, 1e-9))
    print(f'c = {c:8.4g} m/s, Δt = {t_off*1e9:.3f} ns, corresponds to Δd = {t_off * c:.2f} m')
```

```python
fig, ax = plt.subplots(figsize=(4, 2))

for series, label in zip(data, labels):
    ax.plot(lengths, series, label=label)
    f = lambda t, c: (t) * c
    (c,), p_cov = scipy.optimize.curve_fit(f, series*1e-9, 2*lengths, (3e8/2))
    print(f'c = {c:8.4g} m/s')
```
